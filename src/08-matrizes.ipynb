# Matrizes

Você precisa verificar quantas horas de sono uma determinada pessoa dorme
durante o mês, considerando que o mês apresente apenas quatro semanas e que os
dias da semana a considerar devem ser segunda, terça, quarta, quinta e
sexta-feira. Construa uma estrutura de matriz para representar esses dados e
tentar responder as questões apresentadas abaixo.

|    | Seg | Ter | Qua | Qui | Sex |
|----|-----|-----|-----|-----|-----|
| S1 | 8   | 6   | 7   | 8   | 6   |
| S2 | 5   | 5   | 6   | 7   | 8   |
| S3 | 6   | 7   | 8   | 6   | 5   |
| S4 | 7   | 5   | 9   | 7   | 4   |

```python
import enum

class DiaSemana(enum.Enum):
    Seg = 0
    Ter = 1
    Qua = 2
    Qui = 3
    Sex = 4

matriz = [
    [8, 6, 7, 8, 6],
    [5, 5, 6, 7, 8],
    [6, 7, 8, 6, 5],
    [7, 5, 9, 7, 4]
]

# NOTE: Funções Auxiliares

def maximo(lista: list[int]) -> int:
    """Retorna o índice do maior inteiro em uma lista."""
    idx_maximo = 0

    for idx, valor in enumerate(lista):
        if lista[idx_maximo] < valor:
            idx_maximo = idx
    return idx_maximo

def minimo(lista: list[int]) -> int:
    """Retorna o índice do menor inteiro em uma lista."""
    idx_minimo = 0

    for idx, valor in enumerate(lista):
        if lista[idx_minimo] > valor:
            idx_minimo = idx
    return idx_minimo
```

## 1) Qual o dia da semana com maior horário de descanso?

```python
def questao_01():
    for semana, horas_semana in enumerate(matriz):
        idx_maximo_semana = maximo(horas_semana)
        maximo_dia = DiaSemana(idx_maximo_semana)

        print(f"[S{semana + 1}] O dia da semana com maior horário de descanso é a {maximo_dia.name}")

questao_01()
```

## 2) Qual o dia da semana com menor horário de descanso?

```python
def questao_02():
    for semana, horas_semana in enumerate(matriz):
        idx_minimo_semana = minimo(horas_semana)
        minimo_dia = DiaSemana(idx_minimo_semana)
        print(f"[S{semana + 1}] O dia da semana com menor horário de descanso é a {minimo_dia.name}")

questao_02()
```

## 3) Qual a semana com maior horário de descanso?

```python
def questao_03():
    maximos_semanas = [
        horas_semana[maximo(horas_semana)] for horas_semana in matriz
    ]
    semana_com_maior_horario = maximo(maximos_semanas)

    print(f"A semana com o maior horário de descanso é a semana S{semana_com_maior_horario+1}")

questao_03()
```

## 4) Qual a semana com menor horário de descanso?

```python
def questao_04():
    minimos_semanas = [
        horas_semana[minimo(horas_semana)] for horas_semana in matriz
    ]
    semana_com_menor_horario = minimo(minimos_semanas)

    print(f"A semana com o menor horário de descanso é a semana S{semana_com_menor_horario+1}")

questao_04()
```

## 5) Qual o menor tempo de descanso?

```python
def questao_05():
    minimos_semanas = [
        horas_semana[minimo(horas_semana)] for horas_semana in matriz
    ]
    idx_minimo_total = minimo(minimos_semanas)

    print(f"O menor tempo de descanso é {minimos_semanas[idx_minimo_total]}")

questao_05()
```

## 6) Qual o maior tempo de descanso?

```python
def questao_06():
    maximos_semanas = [
        horas_semana[maximo(horas_semana)] for horas_semana in matriz
    ]
    idx_maximo_total = maximo(maximos_semanas)

    print(f"O maior tempo de descanso é {maximos_semanas[idx_maximo_total]}")

questao_06()
```

## 7) Quantos dias da semana tive com o menor tempo de descanso?

```python
def questao_07():
    contador: dict[int, int] = {}

    for horas_semana in (matriz):
        for horas in horas_semana:
            contador.update({ horas: contador.get(horas, 0) + 1 })
    horas_minimas = sorted(contador.keys())[0]
    print(f"{contador[horas_minimas]} dia(s) da semana com menor tempo de descanso")

questao_07()
```

## 8) Quantos dias da semana tive com o maior tempo de descanso?

```python
def questao_08():
    contador: dict[int, int] = {}

    for horas_semana in (matriz):
        for horas in horas_semana:
            contador.update({ horas: contador.get(horas, 0) + 1 })
    horas_maximas = sorted(contador.keys())[-1]
    print(f"{contador[horas_maximas]} dia(s) da semana com maior tempo de descanso")

questao_08()
```

## 9) Qual a percentagem por semana em relação ao total por mês?

```python
def questao_09():
    total = 0
    soma_por_semana: list[int] = []

    for semana in matriz:
        soma = 0
        for horas in semana:
            soma += horas
        total += soma
        soma_por_semana.append(soma)

    for semana, soma_semana in enumerate(soma_por_semana):
        print(f"A porcentagem da S{semana + 1} é de {soma_semana/total*100:.2f}")

questao_09()
```
