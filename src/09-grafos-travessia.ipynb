# Grafos: Travessia

```python
from dataclasses import dataclass


@dataclass(frozen=True)
class Vertice[T]:
    valor: T


class GrafoDirecionado[T]:
    def __init__(self, vertices: list[Vertice[T]]) -> None:
        n = len(vertices)

        self._vertices: list[Vertice[T]] = vertices.copy()
        self._vertices_lookup: dict[Vertice[T], int] = {v:i for i, v in enumerate(self._vertices)}
        self._matriz_adjacencia: list[list[int]] = [[0 for _ in range(n)] for _ in range(n)]

    def adicionar_aresta(self, de: Vertice[T], para: Vertice[T]) -> None:
        de_index = self._vertices_lookup[de]
        para_index = self._vertices_lookup[para]

        self._matriz_adjacencia[de_index][para_index] = 1

    def __getitem__(self, key: Vertice[T]) -> list[Vertice[T]]:
        key_index = self._vertices_lookup[key]
        vizinhos = enumerate(self._matriz_adjacencia[key_index])
        vizinhos = (self._vertices[idx] for idx, value in vizinhos if value > 0)

        return list(vizinhos)
```

## DFS: Depth First Search

Usar DFS para encontrar todos os caminhos entre 2 vértices em um Grafo Direcionado

```python
def buscar_caminhos[T](
    grafo: GrafoDirecionado[T],
    de: Vertice[T],
    para: Vertice[T],
):
    caminhos: list[list[Vertice[T]]] = []

    def dfs(vertice: Vertice[T], caminho: list[Vertice[T]], visitados: set[Vertice[T]]):
        caminho.append(vertice)
        visitados.add(vertice)

        if vertice == para:
            caminhos.append(caminho.copy())
            _ = caminho.pop()
            visitados.remove(vertice)
            return

        vizinhos_nao_visitados = filter(
            lambda v: v not in visitados,
            grafo[vertice],
        )

        for vizinho in vizinhos_nao_visitados:
            dfs(vizinho, caminho, visitados)

        _ = caminho.pop()
        visitados.remove(vertice)

    dfs(de, [], set())
    return caminhos

if True:
    vertices = [
        Vertice(1),
        Vertice(2),
        Vertice(3),
        Vertice(4),
        Vertice(5),
        Vertice(6),
        Vertice(7),
        Vertice(8),
        Vertice(9),
    ]
    grafo = GrafoDirecionado(vertices)
    grafo.adicionar_aresta(vertices[0], vertices[1])
    grafo.adicionar_aresta(vertices[0], vertices[2])
    grafo.adicionar_aresta(vertices[0], vertices[3])
    grafo.adicionar_aresta(vertices[1], vertices[2])
    grafo.adicionar_aresta(vertices[1], vertices[3])
    grafo.adicionar_aresta(vertices[2], vertices[1])
    grafo.adicionar_aresta(vertices[3], vertices[4])
    grafo.adicionar_aresta(vertices[3], vertices[6])
    grafo.adicionar_aresta(vertices[4], vertices[2])
    grafo.adicionar_aresta(vertices[4], vertices[7])
    grafo.adicionar_aresta(vertices[5], vertices[7])
    grafo.adicionar_aresta(vertices[8], vertices[0])
    grafo.adicionar_aresta(vertices[8], vertices[6])

    caminhos = buscar_caminhos(grafo, vertices[0], vertices[4])
    for idx, caminho in enumerate(caminhos):
        print(f"caminho[{idx}] {caminho=}")
```
